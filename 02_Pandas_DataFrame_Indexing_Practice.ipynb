```python
import pandas as pd
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
display(df) 
display(df.loc[["day1","day3","day2"]])
```

```python
import pandas as pd
data = [
    {"calories": 420, "duration": 50},
    {"calories": 380, "duration": 40},
    {"calories": 390, "duration": 45}
]
df = pd.DataFrame(data, index=["day1", "day2", "day3"])
display(df)
```

```python
import pandas as pd
data = [
    [420, 50],
    [380, 40],
    [390, 45]
]
df = pd.DataFrame(data, columns=["calories", "duration"], index=["day1", "day2", "day3"])
display(df)
```
